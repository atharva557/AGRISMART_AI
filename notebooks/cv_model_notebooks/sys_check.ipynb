```python
import sys
print(f"Python version: {sys.version}")
print()

# Try each import separately so if one fails, we still see the rest
try:
    import torch
    print(f" torch: {torch.__version__}")
    print(f"   CUDA available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"   GPU: {torch.cuda.get_device_name(0)}")
    else:
        print("   No GPU detected — training will run on CPU (slower, but fine for this)")
except ImportError as e:
    print(f"torch not installed: {e}")

try:
    import torchvision
    print(f"torchvision: {torchvision.__version__}")
except ImportError as e:
    print(f"torchvision not installed: {e}")

try:
    import sklearn
    print(f"scikit-learn: {sklearn.__version__}")
except ImportError as e:
    print(f"scikit-learn not installed: {e}")

try:
    from PIL import Image
    import PIL
    print(f"Pillow: {PIL.__version__}")
except ImportError as e:
    print(f"Pillow not installed: {e}")
```
